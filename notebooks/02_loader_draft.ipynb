```python
import pandas as pd
commandes = pd.read_excel("../data/raw/commandes.xlsx")
print(commandes.head())
```

```python
lignes = pd.read_excel("../data/raw/lignes.xlsx")

gps = pd.read_excel("../data/raw/gps.xlsx")

print(lignes.head())
print(gps.head())
```

```python
main_table = lignes.merge(
    commandes,
    left_on="Document No_", 
    right_on="No_",
    how="inner"
)
```

```python
print(main_table.shape)
print(main_table.head())
```
